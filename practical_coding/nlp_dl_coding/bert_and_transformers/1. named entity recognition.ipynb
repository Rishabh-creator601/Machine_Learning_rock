```python
from transformers import *
import spacy 
```

```python
text  ="Hi, my name is Ganesh Lokare. I am from Pune. I want to work with Google. I am a student at MIT ADT University. I am currently working on a project related to Named Entity Recognition."
```

## NER with transformers 

=> choose hugging face model from here : [Model List](https://huggingface.co/models?search=named%20entity%20recognition)

```python
ner = pipeline(task="ner", model="NYTK/named-entity-recognition-nerkor-hubert-hungarian")
```

```python
detections =  ner(text,aggregation_strategy="simple")
detections
```

## NER with spacy

```python
!python -m spacy download en_core_web_sm
```

```python
model =  spacy.load("en_core_web_sm")
```

```python
# downgrade ipython to 9.x =>  8.7.0 
docs =  model(text)
spacy.displacy.render(docs, style="ent",jupyter=True)
```
