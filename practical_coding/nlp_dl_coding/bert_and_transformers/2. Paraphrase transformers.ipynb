### Paraphrasing vs. Summarizing  

- **Paraphrasing** → Restating the **whole idea** in your own words, keeping all details.  
  - Example:  
    - Original: *"The sun provides energy to plants through photosynthesis, which helps them grow."*  
    - Paraphrased: *"Plants grow because they use the sun’s energy during photosynthesis."*  

- **Summarizing** → **Shortening** the text by keeping only the **main points**.  
  - Example:  
    - Summary: *"Plants grow using the sun’s energy."*  

---

👉 **Key Difference:**  
- Paraphrasing = same length (different words).  
- Summarizing = much shorter (main idea only).


```python
## JUST SAMPLE SENTENCES 
sentence = "Learning is the process of acquiring new understanding,knowledge, behaviors, skills, values, attitudes, and preferences."
text2 =  "One of the best ways to learn is to teach what you've already learned"
text3 = "Paraphrasing is the process of coming up with someone else's ideas in your own words"
```

```python
def get_response(input_text,num_return_sequences=5,num_beams=5):
  batch = tokenizer([input_text],truncation=True,padding='longest',max_length=60, return_tensors="pt")
  translated = model.generate(**batch,max_length=60,num_beams=num_beams, num_return_sequences=num_return_sequences, temperature=1.5)
  tgt_text = tokenizer.batch_decode(translated, skip_special_tokens=True)
  return tgt_text
```

```python
parrot = Parrot()

phrases =  [sentence, text2, text3]

```

```python
for phrase in phrases:
  print(f"Original Phrase: {phrase}")
  print("Paraphrases:")
  paraphrases = parrot.augment(input_phrase=phrase)
  for para in paraphrases:
    print(para)
  print("\n")
```

**The number accompanied with each sentence is the diversity score. The
higher the value, the more diverse the sentence from the original**
