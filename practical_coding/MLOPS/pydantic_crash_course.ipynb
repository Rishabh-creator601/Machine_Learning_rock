# Pydantic usage 

#### 1. Define a Pydantic model that represents the ideal schema of the data.
- This includes the **expected fields**, their **types**, and **any validation constraints** (e.g., `gt=0` for positive numbers).
#### 2. Instantiate the model with raw input data (usually a dictionary or JSON-like structure).
- Pydantic will automatically **validate** the data and **coerce** it into the correct Python types (if possible).
- If the data doesn't meet the model's requirements, Pydantic raises a `ValidationError`.
#### 3. Pass the validated model object to functions or use it throughout your codebase.
- This ensures that every part of your program works with **clean, type-safe, and logically valid data**.

## 2)  using `Field`

```python
class Patient(BaseModel):
    
    name : str =  Field(max_length=120)
    age : int =  Field(gt=0,lt=120) # gt=greater than , lt =  lower than 

    

p = {"name":"enemy","age":32,"fruits":["apple","mango","banana"]}

print(Patient(**p))
```

## 4)  Field Validator 


=> used to validate fields and do any kind of transformation  <br />
=> must know @staticmethod and @classmethod before it <br />
=>  it is simply Patient.transform_name(name) <br />
=>  a concept of `mode='after'` and `mode='before'` exists  in field validator 


static method : [static method ](https://www.geeksforgeeks.org/python/python-staticmethod/) | 
class method :  [class method](https://www.geeksforgeeks.org/python/classmethod-in-python/)

```python
class Patient(BaseModel):
    
    name : str 
    age : int 
    disease_name :  Optional[str] = None # default
    fruits : list[str]
    
    
    @field_validator("name",mode="before")
    @classmethod
    def transform_name(cls,data):
        data =  data.strip().upper()
        return data
    


p = {"name":"enemy","age":32,"fruits":["apple","mango","banana"]}

print(Patient(**p))
```

## 5) Model Validator 

=> It is applied to the whole model rather than to the specific instance. 


```python
class Patient(BaseModel):
    
    name : str 
    age : int 
    disease_name :  Optional[str] = None # default
    fruits : list[str]
    
    
    
    
    @model_validator(mode="after")
    def print_something(cls,model):
        
        if model.age >  60 and "mango"  in model.fruits :
            raise ValueError("age must be less than 60 ")
        return model
    
    


p = {"name":"enemy","age":9,"fruits":["apple","mango","banana"]}

print(Patient(**p))
```

## 6) Computed Field 

```python
class Patient(BaseModel):
    
    name : str 
    age : int 
    disease_name :  Optional[str] = None # default
    fruits : list[str]
    
    
    
    @computed_field
    @property
    def calculate_cat(self) -> str:
        
        cat = "old"  if self.age  > 20 else  "teenage "
        return cat

    
    


p = {"name":"enemy","age":70,"fruits":["apple","mango","banana"]}

patient =  Patient(**p)


print(patient)
print(patient.calculate_cat)


```

## 7)  Serialization 

```python
class Patient(BaseModel):
    
    name : str 
    age : int 
    disease_name :  Optional[str] = None # default
    fruits : list[str]
    
    


p = {"name":"enemy","age":70,"fruits":["apple","mango","banana"]}

patient =  Patient(**p)


print(patient.model_dump())
print(patient.model_dump(include=["name","age"]))

print(patient.model_dump(exclude_unset=True)) # ignore the unset values 

```
