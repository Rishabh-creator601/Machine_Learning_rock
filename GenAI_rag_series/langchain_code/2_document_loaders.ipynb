```python
from langchain_community.document_loaders import TextLoader,PyMuPDFLoader,PyPDFLoader,CSVLoader
from langchain_core.output_parsers import StrOutputParser
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from dotenv import load_dotenv

load_dotenv()
```

## TABLE OF CONTENTS 

1)  Text Loader 
2)  PyPDFLoader
3)  CSVLoader 


#### theory part 
**Length is counted till .load() that does not include [0].page_content part**  <br />





## 1) Text Loader 

```python
model  = ChatOpenAI()
Parser =  StrOutputParser()
prompt =  PromptTemplate(
    template="generate 5 important points from {topic}",
    input_variables=['topic']
)
loader =  TextLoader("assets/file_dopamine.txt").load()[0].page_content

chain =  prompt |  model | Parser 




print("-"*10 +  "5 points" +  "-"*10)
print(chain.invoke({'topic':loader}))


print("-"*10 +  "TEXT CONTENT" +  "-"*10)
print(loader)




```

## 2) PyPDF loader 

```python
model  = ChatOpenAI()
Parser =  StrOutputParser()
prompt =  PromptTemplate(
    template="generate 5 important points from {topic}",
    input_variables=['topic']
)
loader = PyPDFLoader("assets/personality_types.pdf").load()[0].page_content

chain =  prompt |  model | Parser 




print("-"*10 +  "5 points" +  "-"*10)
print(chain.invoke({'topic':loader}))

print("-"*10 +  "TEXT CONTENT" +  "-"*10)
print(loader)


```

## 3)  CSV Loader 

```python
model  = ChatOpenAI()
Parser =  StrOutputParser()
prompt =  PromptTemplate(
    template="generate 5 important points from {topic}",
    input_variables=['topic']
)
loader = CSVLoader(file_path="assets/students_data.csv",encoding="utf-8-sig").load()

chain =  prompt |  model | Parser 






print("-"*10 +  "5 points" +  "-"*10)
print(chain.invoke({'topic':loader}))

print("-"*10 +  "TEXT CONTENT" +  "-"*10)
print(loader)
```
