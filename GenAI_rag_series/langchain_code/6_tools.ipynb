## 1)  Web Search (DuckDuckGoRun)

```python
search = DuckDuckGoSearchRun()

web_search =  Tool(
    name= "Web Search",
    func=search.run,
    description="search the relevant articles and topic discussions"
)

query = "top 5 important python packages ? "

ans =  web_search.invoke(query)
print(ans)
```

## 2)  Shell Tool 

```python
shell =  ShellTool()


result = shell.invoke({"commands":["dir"]})
print(result)
```

## 3)  Python REPL Tool 

```python
pythonTool =  PythonREPLTool()

query = """

def multiply(a,b):
    return 'result is ' +  str(a*b) 


print(multiply(3,4))

"""


result =  pythonTool.invoke({"query":query})
print(result)
```
